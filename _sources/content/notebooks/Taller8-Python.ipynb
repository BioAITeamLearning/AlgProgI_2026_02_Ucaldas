## Guía rápida

**Ciclo `while`.** Repite el bloque mientras la condición sea verdadera. Con `break` se sale del ciclo en cualquier momento.

```python
contador = 1
while contador <= 3:
    print(contador)
    contador += 1        # sin esta línea el ciclo nunca termina

while True:              # ciclo infinito: se sale con break
    texto = input("Escriba 'salir': ")
    if texto == "salir":
        break
```

**Ciclo `for` y `range`.** `range(inicio, fin, paso)` genera números desde `inicio` hasta `fin - 1`.

| Instrucción | Valores que toma `i` |
| --- | --- |
| `for i in range(5):` | 0, 1, 2, 3, 4 |
| `for i in range(1, 6):` | 1, 2, 3, 4, 5 |
| `for i in range(0, 10, 2):` | 0, 2, 4, 6, 8 |
| `for i in range(5, 0, -1):` | 5, 4, 3, 2, 1 |

El `for` también recorre directamente un texto (letra por letra) o una lista (elemento por elemento):

```python
for letra in "hola":
    print(letra)
```

**Listas.** Guardan varios valores en una sola variable. Las posiciones empiezan en 0.

```python
frutas = ["mango", "pera"]
frutas.append("uva")      # agrega al final -> ["mango", "pera", "uva"]
print(frutas[0])          # mango
print(len(frutas))        # 3
for f in frutas:          # recorre los elementos
    print(f)
for i in range(len(frutas)):   # recorre las posiciones
    print(i, frutas[i])
```

**`match` (el *switch* de Python).** Compara un valor contra varios casos. `case _` es el caso por defecto y `|` une varios valores en un mismo caso.

```python
match opcion:
    case 1:
        print("Uno")
    case 2 | 3:
        print("Dos o tres")
    case _:
        print("Otro valor")
```

**Funciones.** Se definen con `def`. Una función puede **no retornar** nada (solo hace algo, como imprimir) o **retornar** un valor con `return`.

```python
def saludar(nombre):          # no retorna: solo imprime
    print("Hola,", nombre)

def cuadrado(x):              # retorna un valor
    return x * x

saludar("Ana")
resultado = cuadrado(4)       # resultado vale 16
```
