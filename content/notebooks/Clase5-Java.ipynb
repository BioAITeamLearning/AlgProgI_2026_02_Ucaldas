### Ejemplos `if`

#### Ejemplo 1:

Realizar un programa que me diga si un numero $X$ es mayor que $5$

```java
// Variable scn de tipo scanner
Scanner scn = new Scanner(System.in);

// Leer número
System.out.println("Ingrese un número:");
float x = scn.nextFloat();

// Acción
if (x > 5) {
    System.out.println("x es mayor que 5");
}
```

#### Ejemplo 2:

Realizar un programa que me diga si una persona es mayor de edad e imprima el mensaje `"Mientras usted viva en esta casa se hará lo que yo diga"`, en caso de que viva en la casa de sus padres y sea mayor de edad. Debe pedir la edad del usuario y preguntar si vive o no en la casa de sus padres.

*Implementación de `if` anidados. Los `if` anidados son estructuras que establecen condiciones consecutivas donde se encandenan, esto tiene diferentes aplicaciones desde hacer validaciones hasta generar filtros de condición.*

```java
// Variable para leer la edad
Scanner scn = new Scanner(System.in);

// Variable para leer si o no
Scanner scin = new Scanner(System.in);

// Leer edad
System.out.println("Ingrese la edad:");
int edad = scn.nextInt();

// Leer si o no
System.out.println("Vive en casa (si/no):");
String casa = scin.nextLine();

// Acción con if anidados
if (edad>0) {
  if(edad>=18) {
    System.out.println("Es mayor de edad");
    if (casa.equals("si")) {
      System.out.println("Mientras usted viva en esta casa se hará lo que yo diga");
    }
    else{
      System.out.println("puede hacer lo que le de la gana en su casa");
      }
  }
  else{
    System.out.println("usted es un bebe");
  }
}
else{
  System.out.println("aprenda a escribir edades");
}
```

#### Ejemplo 7:

*`else if` es una alternativa al uso de los `if` anidados y la estructura `else`, este se activa como una condición alternativa si el resultado es falso en la condición inicial, esta funciona como un `else` pero iterativo y consecutivamente se cierra con `else`.*

#### Ejemplo 11:

Realizar un programa que cuente los días que faltan para el fin de semana. Tenga en cuenta que:

    Domingo = 1 (Fin de Semana)
    Lunes = 2
    Martes = 3
    Miercoles = 4
    Jueves = 5
    Viernes = 6
    Sabado = 7 (Fin de Semana)

```java
// Variable para leer dia de la semana
Scanner scn = new Scanner(System.in);

// Leer dia
System.out.println("Ingrese el dia de la semana: ");
int dia = scn.nextInt();

// Variable para contar los dias que faltan
int diasHastaFinSemana = 0;

// Lógica
if (dia > 1 && dia < 7) {
    if (dia < 6) {
        diasHastaFinSemana++;
        if (dia < 5) {
           diasHastaFinSemana++;
           if (dia < 4) {
                diasHastaFinSemana++;
                if (dia < 3) {
                    diasHastaFinSemana++;
                }
            }
        }
    }
    System.out.println("Dias restantes hasta el fin de semana: " + diasHastaFinSemana);
} else if (dia == 1 || dia == 7) {
    System.out.println("Fin de semana");
} else {
    System.out.println("La semana solo tiene 7 dias");
}
```

## Ejemplos Switch Case

Usa `switch` cuando necesitas comparar **una misma variable (o expresión)** contra varios valores exactos. Puede ser de tipo `int`, `char` o `String` (no admite `double`, `boolean` ni rangos como `x > 5`).

- Cada `case` es un valor posible.
- `break` termina el `switch`.
- `default` se ejecuta cuando ningún `case` coincide.

#### Ejemplo 12: `switch` con `String`

Realizar un programa que pida al usuario un tipo de vehiculo (`Carro`, `Camion`, `Moto`) y que imprima a que hora del dia puede circular dependiendo del tipo ingresado. Si el usuario ingresa un tipo de vehiculo distinto debe mostrar un mensaje indicando el error.

```java
Scanner scn = new Scanner(System.in);
```

```java
// Leer el vehiculo (String)
System.out.println("En que vehiculo se desplaza (Carro/Camion/Moto)");
String tipoVehiculo = scn.nextLine();
```

```java
// Lógica
switch (tipoVehiculo) {
   case "Carro":
      System.out.println("Puedes pasar de 00:00 a 08:00");
      break;
   case "Camion":
      System.out.println("Puedes pasar de 08:00 a 16:00");
      break;
   case "Moto":
      System.out.println("Puedes pasar de 16:00 a 24:00");
      break;
   default:
      System.out.println("No se puede pasar con un " + tipoVehiculo);
      break;
}
```

#### Ejemplo 13: `switch` con `int`

Realizar un programa que pida al usuario el plan de almacenamiento en la nube que desea (`1`, `2` o `3`) y que imprima sus características. Si ingresa otro número debe indicar que el plan no existe.

```java
Scanner scn = new Scanner(System.in);
```

```java
// Leer el plan (int)
System.out.println("Elige tu plan de almacenamiento (1 = Gratis, 2 = Estándar, 3 = Premium)");
int plan = scn.nextInt();
```

```java
// Lógica
switch (plan) {
    case 1:
        System.out.println("Plan Gratis: 15 GB por $0 al mes");
        break;
    case 2:
        System.out.println("Plan Estándar: 100 GB por $9.900 al mes");
        break;
    case 3:
        System.out.println("Plan Premium: 1 TB por $29.900 al mes");
        break;
    default:
        System.out.println("El plan " + plan + " no existe");
        break;
}
```

#### Ejemplo 14: ¿Qué hace `break`?

`break` le dice a Java que **salga del `switch`**. Si lo olvidas, Java no se detiene: sigue ejecutando todos los `case` que vienen después, aunque no coincidan (esto se llama *fall-through*, o "caída").

Tomemos el `switch` del ejemplo anterior y quitémosle todos los `break`. Elegimos el plan 2:

```java
// SIN break
int planElegido = 2;
```

```java
switch (planElegido) {
    case 1:
        System.out.println("Plan Gratis: 15 GB por $0 al mes");
    case 2:
        System.out.println("Plan Estándar: 100 GB por $9.900 al mes");
    case 3:
        System.out.println("Plan Premium: 1 TB por $29.900 al mes");
    default:
        System.out.println("El plan " + planElegido + " no existe");
}
```

Se eligió el plan 2 y se imprimieron **tres** mensajes: Java entró al `case 2` y, como no había `break`, siguió con el `case 3` e incluso con el `default`.

Ahora el mismo `switch` **con** `break` en cada `case`:

```java
// CON break
int opcionElegida = 2;
```

```java
switch (opcionElegida) {
    case 1:
        System.out.println("Plan Gratis: 15 GB por $0 al mes");
        break;
    case 2:
        System.out.println("Plan Estándar: 100 GB por $9.900 al mes");
        break;
    case 3:
        System.out.println("Plan Premium: 1 TB por $29.900 al mes");
        break;
    default:
        System.out.println("El plan " + opcionElegida + " no existe");
        break;
}
```

#### Ejemplo 15: `switch` con `char` y varios `case` juntos

Realizar un programa que pida al usuario confirmar una operación con `S` (sí) o `N` (no). Debe aceptar la letra en mayúscula o en minúscula. Si ingresa otra letra debe indicar que la respuesta no es válida.

Para leer un `char` se usa `scn.next().charAt(0)`: toma el primer carácter de lo que escribe el usuario. Y cuando varios `case` deben hacer lo mismo se escriben seguidos, **sin `break` entre ellos** (a propósito); el `break` va al final del grupo.

```java
Scanner scn = new Scanner(System.in);
```

```java
// Leer la respuesta (char)
System.out.println("¿Desea confirmar la operación? (S/N)");
char respuesta = scn.next().charAt(0);
```

```java
// Lógica
switch (respuesta) {
    case 's':
    case 'S':
        System.out.println("Operación confirmada");
        break;
    case 'n':
    case 'N':
        System.out.println("Operación cancelada");
        break;
    default:
        System.out.println("Respuesta no válida, use S o N");
        break;
}
```

#### Ejemplo 16: una expresión dentro del `switch`

Entre los paréntesis del `switch` no solo puede ir una variable: también una expresión. Realizar un programa que pida la nota de un estudiante (de $0.0$ a $5.0$) y muestre su desempeño según la parte entera de la nota: `5` es Excelente, `4` es Bueno, `3` es Aceptable y cualquier otro valor es Insuficiente.

```java
Scanner scn = new Scanner(System.in);
```

```java
// Leer la nota (double)
System.out.println("Ingrese la nota (0.0 a 5.0): ");
double nota = scn.nextDouble();
```

```java
// (int) nota descarta los decimales: 4.7 se convierte en 4
switch ((int) nota) {
    case 5:
        System.out.println("Excelente");
        break;
    case 4:
        System.out.println("Bueno");
        break;
    case 3:
        System.out.println("Aceptable");
        break;
    default:
        System.out.println("Insuficiente");
        break;
}
```
