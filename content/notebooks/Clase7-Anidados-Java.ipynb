## Ejemplos de ciclos anidados

Un ciclo anidado es un ciclo dentro de otro. Por cada vuelta del ciclo de afuera, el ciclo de adentro se ejecuta completo.

**Ejemplo 1.**

Realiza un programa que imprima las tablas de multiplicar desde la del $1$ hasta la del número que ingrese el usuario. Cada tabla debe llegar hasta $\times 10$.

El ciclo de afuera recorre las tablas y el ciclo de adentro recorre los multiplicadores de $1$ a $10$.

```java
Scanner input = new Scanner(System.in);
System.out.print("Ingrese hasta qué tabla desea ver: ");
int n = input.nextInt();

for (int tabla = 1; tabla <= n; tabla++) {
    System.out.println("Tabla del " + tabla);
    for (int i = 1; i <= 10; i++) {
        System.out.println(tabla + " x " + i + " = " + (tabla * i));
    }
    System.out.println();
}
```

**Ejemplo 2.**

Un profesor de ciencia de datos quiere calcular el promedio de notas de sus estudiantes. El usuario debe ingresar cuántos estudiantes hay y cuántas notas tiene cada uno. Luego el programa pide las notas y muestra el promedio de cada estudiante.

El ciclo de afuera recorre los estudiantes y el ciclo de adentro recorre las notas de ese estudiante. Fíjate que la suma se reinicia en $0$ para cada estudiante.

```java
Scanner input = new Scanner(System.in);
System.out.print("Número de estudiantes: ");
int estudiantes = input.nextInt();
System.out.print("Notas por estudiante: ");
int notas = input.nextInt();

for (int e = 1; e <= estudiantes; e++) {
    double suma = 0;
    for (int k = 1; k <= notas; k++) {
        System.out.print("Estudiante " + e + ", nota " + k + ": ");
        suma += input.nextDouble();
    }
    System.out.println("Promedio del estudiante " + e + ": " + suma / notas);
}
```

**Ejemplo 3.**

Los ciclos anidados no tienen que ser del mismo tipo. Realiza un programa que, para cada número desde $1$ hasta $n$, imprima la suma de sus dígitos. El usuario ingresa $n$.

Por ejemplo, para $12$ la suma de dígitos es $1 + 2 = 3$.

El ciclo `for` recorre los números y el ciclo `while` separa los dígitos de cada uno: `% 10` obtiene el último dígito y `/ 10` lo elimina.

```java
Scanner input = new Scanner(System.in);
System.out.print("Ingrese n: ");
int n = input.nextInt();

for (int numero = 1; numero <= n; numero++) {
    int temp = numero;
    int sumaDigitos = 0;
    while (temp > 0) {
        sumaDigitos += temp % 10;
        temp /= 10;
    }
    System.out.println(numero + " -> suma de dígitos: " + sumaDigitos);
}
```
